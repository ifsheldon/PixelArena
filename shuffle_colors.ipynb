```python
from pixel_arena.dataset_utils.celeb_a_mask_hq import LABEL_COLORS, LABELS
import numpy as np
import pickle
import os
from label_encoding_palette import plot_label_palette
```

```python
# shuffle the colors
reshuffle = False
pkl_store_path = "./label_palettes/shuffled_colors.pkl"
label_palette_path = "./label_palettes/shuffled_colors.png"
if not os.path.exists(pkl_store_path) or reshuffle:
    arange = np.arange(len(LABEL_COLORS))
    np.random.shuffle(arange)
    shuffled_idx = arange.tolist()
    shuffled_colors = [LABEL_COLORS[i] for i in shuffled_idx]
    print(shuffled_colors)

    with open(pkl_store_path, "wb") as f:
        pickle.dump(shuffled_colors, f)
else:
    with open(pkl_store_path, "rb") as f:
        shuffled_colors = pickle.load(f)
        print(f"Loaded shuffled colors: {shuffled_colors}")

plot_label_palette(LABELS, shuffled_colors, columns=4, save_path=label_palette_path)
```

```python
from pixel_arena.generation.gemini import batch_gen_mask
from pathlib import Path
```

```python
attempts = 5

await batch_gen_mask(
    image_dir=Path("./eval-set/images-150"),
    output_dir=Path("./results/gemini-pro-150-shuffled-label-colors"),
    color_palette_path=Path(label_palette_path),
    attempts=attempts,
    label_colors=shuffled_colors,
)
```

```python
from convert_masks import batch_process_images

batch_process_images(
    "./results/gemini-pro-150-shuffled-label-colors",
    label_colors=shuffled_colors,
    attempts=attempts,
    force=True,
)
```
